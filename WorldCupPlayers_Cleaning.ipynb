```python
import plotly.express as px

team_count = df.groupby("Team Initials").size().reset_index(name="Player Count")

fig = px.bar(
    team_count,
    x="Team Initials",
    y="Player Count",
    color="Player Count",
    text="Player Count",
    title="🌍 Players by Team"
)

fig.update_layout(template="plotly_dark", title_x=0.5)

fig.show()
```

```python
import plotly.express as px

treemap_data = df.groupby(["Team Initials", "Position"]).size().reset_index(name="Count")

fig = px.treemap(
    treemap_data,
    path=["Team Initials", "Position"],
    values="Count",
    color="Count",
    color_continuous_scale="Turbo",
    title="⚽ Player Distribution by Team and Position"
)

fig.update_layout(
    template="plotly_dark",
    title_x=0.5
)

fig.show()
```

```python
import plotly.express as px

heatmap_data = df.groupby(["Position", "Shirt Number"]).size().reset_index(name="Count")

fig = px.density_heatmap(
    heatmap_data,
    x="Shirt Number",
    y="Position",
    z="Count",
    color_continuous_scale="Turbo",   # Try Turbo, Plasma, Inferno, Viridis
    text_auto=True,                   # Shows the count inside each cell
    title="👕 Shirt Numbers Used for Each Player Position"
)

fig.update_layout(
    template="plotly_dark",
    title={
        "text": "👕 Shirt Numbers Used for Each Player Position",
        "x": 0.5,
        "font": dict(size=24)
    },
    xaxis_title="Shirt Number",
    yaxis_title="Player Position",
    font=dict(size=14),
    coloraxis_colorbar=dict(
        title="Players"
    )
)

fig.update_traces(
    hovertemplate="<b>Position:</b> %{y}<br>"
                  "<b>Shirt Number:</b> %{x}<br>"
                  "<b>Players:</b> %{z}<extra></extra>"
)

fig.show()
```

```python
team_players = (
    df.groupby("Team Initials")
      .size()
      .reset_index(name="Player Count")
      .sort_values("Player Count", ascending=True)
)

fig = px.bar(
    team_players,
    x="Player Count",
    y="Team Initials",
    orientation="h",
    color="Player Count",
    color_continuous_scale="Viridis",
    text="Player Count",
    title="⚽ Number of Players Representing Each Team"
)

fig.update_layout(
    template="plotly_dark",
    title={
        "text": "⚽ Number of Players Representing Each Team",
        "x": 0.5,
        "font": dict(size=24)
    },
    xaxis_title="Number of Players",
    yaxis_title="Team Initials",
    font=dict(size=14),
    height=700
)

fig.update_traces(
    textposition="outside",
    hovertemplate="<b>Team:</b> %{y}<br><b>Players:</b> %{x}<extra></extra>"
)

fig.show()
```

```python
import plotly.express as px

shirt_count = (
    df.groupby("Shirt Number")
      .size()
      .reset_index(name="Player Count")
      .sort_values("Shirt Number")
)

fig = px.bar(
    shirt_count,
    x="Shirt Number",
    y="Player Count",
    color="Player Count",
    text="Player Count",
    color_continuous_scale="Inferno",
    title="👕 Most Commonly Used Shirt Numbers"
)

fig.update_traces(
    textposition="outside",
    marker_line_color="white",
    marker_line_width=1.5,
    hovertemplate="<b>Shirt Number:</b> %{x}<br><b>Players:</b> %{y}<extra></extra>"
)

fig.update_layout(
    template="plotly_dark",
    title={
        "text": "👕 Most Commonly Used Shirt Numbers",
        "x": 0.5,
        "font": dict(size=26)
    },
    xaxis_title="Shirt Number",
    yaxis_title="Number of Players",
    width=1200,
    height=650,
    font=dict(size=14),
    coloraxis_colorbar=dict(title="Players")
)

fig.show()
```
