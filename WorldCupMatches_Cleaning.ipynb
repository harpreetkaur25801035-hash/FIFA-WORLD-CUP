```python
stage_count = df["Stage"].value_counts().reset_index()
stage_count.columns = ["Stage", "Count"]

fig = px.bar(
    stage_count,
    x="Stage",
    y="Count",
    color="Count",
    color_continuous_scale="Viridis",
    title="🏆 Number of Matches in Each Stage",
    template="plotly_dark",
    text="Count",
    
)

fig.update_layout(
    title_x=0.5,
    xaxis_tickangle=-45
)

fig.show()
```

```python
import plotly.express as px

fig = px.scatter(
    df,
    x="Home Team Goals",
    y="Away Team Goals",
    title="⚽ Home Team Goals vs Away Team Goals",
    color_discrete_sequence=["#331fb4"]
)

fig.update_traces(
    marker=dict(
        size=12,
        opacity=0.8,
        line=dict(width=1.5, color="black")
    )
)

fig.update_layout(
    template="plotly_dark",
    title_font_size=24,
    title_x=0.5,
    xaxis_title="Home Team Goals",
    yaxis_title="Away Team Goals",
    width=900,
    height=550
)

fig.show()
```

```python
import plotly.express as px

data = df.groupby("Year")[["Home Team Goals","Away Team Goals"]].mean().reset_index()

fig = px.line(
    data,
    x="Year",
    y=["Home Team Goals","Away Team Goals"],
    title="⚽ Average Home Team Goals vs Away Team Goals by Year"
)

fig.update_traces(line=dict(width=4))

fig.data[0].line.color = "#90EE90"   # Light Green
fig.data[1].line.color = "#FFD700"   # Yellow

fig.update_layout(
    template="plotly_dark",
    title_x=0.5,
    width=950,
    height=550,
    xaxis_title="Year",
    yaxis_title="Average Goals"
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px
import numpy as np

# Create Match Outcome column
df['Match Outcome'] = np.where(
    df['Home Team Goals'] > df['Away Team Goals'],
    'Home Win',
    np.where(
        df['Home Team Goals'] < df['Away Team Goals'],
        'Away Win',
        'Draw'
    )
)

# Count matches by outcome
outcome_counts=(
    df.groupby('Match Outcome', as_index=False)
    .size()
    .rename(columns={'size': 'Match Count'})
)
# Create donut chart
fig = px.pie(
    outcome_counts,
    names='Match Outcome',
    values='Match Count',
    hole=0.45,  # Makes it a donut chart
    title='Distribution of Match Outcomes',
    color='Match Outcome',
    color_discrete_map={
        'Home Win': "#e74096",
        'Away Win': "#27c4d6",
        'Draw': "#f0e570"
    }
)

# Show percentage and label
fig.update_traces(
    textposition='inside',
    textinfo='percent+label'
)
# Improve layout
fig.update_layout(
    template='plotly_dark',
    legend_title='Match Outcome'
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px

# Calculate total goals
df["Total Goals"] = df["Home Team Goals"] + df["Away Team Goals"]

# Create match description
df["Match Description"] = (
    df["Year"].astype(str) + " - " +
    df["Home Team Name"] + " vs " +
    df["Away Team Name"]
)

# Get top 10 highest scoring matches
top_matches = df.sort_values(by="Total Goals", ascending=False).head(10)

# Plotly Horizontal Bar Chart
fig = px.bar(
    top_matches,
    x="Total Goals",
    y="Match Description",
    color="Year",
    orientation="h",
    text="Total Goals",
    title="Top 10 Highest Scoring FIFA World Cup Matches",
    color_continuous_scale="Blues"
)

fig.update_traces(textposition="outside")
fig.update_layout(
    xaxis_title="Total Goals",
    yaxis_title="Match (Year - Home vs Away)",
    template="plotly_dark",
    yaxis=dict(categoryorder="total ascending")
)

fig.show()
```
