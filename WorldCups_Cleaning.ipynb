```python

import plotly.express as px
import pandas as pd
df = pd.read_csv("WorldCups.csv")

df_new = df[["Year", "GoalsScored", "MatchesPlayed"]]

df_new = df_new.melt(
    id_vars="Year",
    value_vars=["GoalsScored", "MatchesPlayed"],
    var_name="Metric",
    value_name="Value"
)

fig = px.line(
    df_new,
    x="Year",
    y="Value",
    color="Metric",
    markers=True,color_discrete_map={
    "GoalsScored": "#EAEF40",      # Gold
    "MatchesPlayed": "#A7E98B"     # Deep Sky Blue
    },

    
    title="Goals Scored and Matches Played Over the Years"
)

fig.update_layout(
    template="plotly_dark",
    xaxis_title="Year",
    yaxis_title="Count",
    title_x=0.5
)

fig.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=df["Year"],
    y=df["QualifiedTeams"],
    mode="markers+text",
    marker=dict(
        size=14,
        color="#FF00C3",
        line=dict(color="white", width=2)
    ),
    text=df["QualifiedTeams"],
    textposition="top center"
))

for x, y in zip(df["Year"], df["QualifiedTeams"]):
    fig.add_shape(
        type="line",
        x0=x,
        x1=x,
        y0=0,
        y1=y,
        line=dict(color="#00E5FF", width=3)
    )

fig.update_layout(
    template="plotly_dark",
    title="Qualified Teams in FIFA World Cup",
    title_x=0.5,
    xaxis_title="Year",
    yaxis_title="Qualified Teams"
)

fig.show()
```

```python
import plotly.express as px

country_matches = df.groupby("Country").size().reset_index(name="Matches")

country_matches = country_matches.sort_values(by="Matches", ascending=False)

fig = px.bar(
    country_matches,
    x="Country",
    y="Matches",
    color="Matches",
    color_continuous_scale="Tealgrn",
    title="Which Country Hosted the Most World Cup Matches?",
    template="plotly_dark",
    text="Matches"
)

fig.update_traces(textposition="outside")

fig.update_layout(
    xaxis_title="Host Country",
    yaxis_title="Number of Matches",
    showlegend=False,
    xaxis_tickangle=-45
)

fig.show()
```

```python
import plotly.express as px

winner_count = df.groupby("Winner").size().reset_index(name="World Cup Titles")

winner_count = winner_count.sort_values(
    by="World Cup Titles",
    ascending=False
)

fig = px.bar(
    winner_count,
    x="World Cup Titles",
    y="Winner",
    orientation="h",
    color="World Cup Titles",
    color_continuous_scale="YlOrBr",
    template="plotly_dark",
    text="World Cup Titles",
    title="Countries with the Most FIFA World Cup Titles"
)

fig.update_traces(textposition="outside")

fig.update_layout(
    title_x=0.5,
    xaxis_title="Number of World Cup Titles",
    yaxis_title="Country",
    showlegend=False,
    font=dict(size=15)
)

fig.show()
```
